## 1. Your first program: `print()` and comments

A program is a list of instructions for the computer. Python reads them **from top to bottom**, one line at a time.

`print()` is a **function** that shows something on the screen. Whatever you put inside the parentheses is called the **argument**.

- Text must be inside quotes (`"..."` or `'...'`). Text is called a **string**.
- Numbers go without quotes.

A **comment** starts with `#`. Python ignores it completely. Comments are notes for humans.

### 📚 From class

````python
print("hello")
````

````python
# This line is a comment. Python skips it.
print("Hello World")      # a comment can also go at the end of a line
````

**What's the difference between these two?** Run them and compare.

````python
print(5 + 3)        # Python calculates this
print("5 + 3")      # Python just prints the text
````

A number without quotes is **data**: Python can calculate with it. The same characters inside quotes are **text**: Python just shows them.

### 🆕 More examples

`print()` can take several values separated by commas. By default it puts a **space** between them (`sep`) and a **new line** at the end (`end`). You can change both:

````python
print("Python", "is", "fun")
print("Python", "is", "fun", sep="-")
print("Loading", end="...")
print("done!")
````

Quotes can be tricky. Use double quotes if your text contains a single quote (and the other way around). `\n` means "new line" and `\"` means "a real double quote".

````python
print("It's a beautiful day")
print('She said "hello"')
print("Line one\nLine two")
print("A backslash: \\")
````

Triple quotes let you write text over **many lines**:

````python
poem = """Roses are red,
Violets are blue,
Python is awesome,
And so are you."""
print(poem)
````

> 💡 **Remember:** Python is **case-sensitive**. `print` works, but `Print` and `PRINT` do not.

## 2. Variables

A **variable** is a name that points to a value stored in memory. Think of it as a **labelled box**: the label is the name, the box holds the value.

We create a variable with the **assignment operator** `=`. It means: *"evaluate the right side, then store the result under the name on the left."*

Unlike Java or C#, Python does **not** need you to declare the type first. You just assign.

### 📚 From class

````python
x = 4
print(x)

x, y, z = 1, 2, 3        # assign three variables at once
print(x, y, z)

x = y = z = 50           # give three names the same value
print(x, y, z)
````

Reassigning is the same syntax. The new value **replaces** the old one, and the right side can use the old value:

````python
spam_amount = 0
print(spam_amount)

spam_amount = spam_amount + 4     # take the old value, add 4, store it back
print(spam_amount)
````

### 🆕 More examples

````python
score = 10
print("Score:", score)

score = score + 5
print("Score after a bonus:", score)

score += 5                # shortcut for: score = score + 5
print("Score after another bonus:", score)

score -= 3                # also: *=  /=  //=  %=
print("Score after a penalty:", score)
````

Swapping two variables is easy in Python:

````python
a = "left"
b = "right"
a, b = b, a
print(a, b)
````

### Naming rules
| Rule | Good | Bad |
|---|---|---|
| Letters, digits and `_` only | `player_score`, `age2` | `my-score`, `my score` |
| Can't start with a digit | `level1` | `1level` |
| Case-sensitive | `Age` and `age` are different | |
| Can't be a Python keyword | `class_name` | `class`, `if`, `for` |
| Use `snake_case` by convention | `first_name` | `FirstName` (this style is for classes) |

> ⚠️ **Common mistake:** Do not name a variable after a built-in function like `sum`, `list`, `str`, `max`, or `print`. Python lets you, but then the built-in stops working!

````python
numbers = [4, 8, 15]
print(sum(numbers))        # works: sum is the built-in function

sum = 100                  # oops, we just replaced the built-in with a number
````

````python
print(sum(numbers))        # now this fails
````

````python
del sum                    # delete our variable so the built-in works again
print(sum(numbers))
````

## 3. Data types

Every value has a **type**. The four basic ones:

| Type | Name | Examples | Meaning |
|---|---|---|---|
| `int` | integer | `5`, `-3`, `0` | whole numbers |
| `float` | floating point | `5.0`, `3.14`, `-0.5` | numbers with a decimal point |
| `str` | string | `"hello"`, `'a'` | text |
| `bool` | boolean | `True`, `False` | yes / no |

`type()` tells you what kind of value you have. It is one of the most useful built-ins when something behaves strangely.

### 📚 From class

````python
x = 5
y = 5.0
a = 'a'
b = 'hello'
c = True

print(type(x))
print(type(y))
print(type(a))
print(type(b))
print(type(c))
print(type(str(y)))      # str() converts to text, so this is a string
````

Python is **dynamically typed**: the *variable* has no fixed type; the *value* does. The same name can hold different types over time (not recommended, but allowed).

````python
thing = 10
print(type(thing))
thing = "ten"
print(type(thing))
````

### 🆕 More examples

Types decide what operations make sense. `+` adds numbers but **joins** strings. `*` multiplies numbers but **repeats** strings:

````python
print(3 + 4)
print("3" + "4")
print("ha" * 3)
print(3 * 4)
````

You can't mix them freely. Run this and read the error (we'll learn to read errors in section 10):

````python
age = 15
print("I am " + age + " years old")
````

Fix 1: convert the number to a string with `str()`. Fix 2 (better): use an **f-string**, which puts values straight into text. Put an `f` before the quote and write variables inside `{ }`.

````python
age = 15
name = "Ayesha"
print("I am " + str(age) + " years old")
print(f"{name} is {age} years old")
print(f"Next year {name} will be {age + 1}")      # you can calculate inside { }
````

> 💡 **Fun fact:** Python integers have no size limit, so `2 ** 200` works fine. In Java or C# an `int` overflows at about 2 billion.

````python
print(2 ** 200)
````

## 6. Booleans and comparisons

A `bool` has only two values: `True` and `False` (capital first letter!). We usually get booleans from **comparison operators**, which answer yes/no questions.

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `<` / `>` | less than / greater than |
| `<=` / `>=` | less than or equal / greater than or equal |

### ⚠️ `=` versus `==`
- `=` **stores** a value: `x = 5` means "put 5 into x".
- `==` **asks a question**: `x == 5` means "is x equal to 5?"

### 📚 From class

````python
age = 20
print(age > 18)
print(age == 20)
print(age != 20)
print(type(age > 18))
````

````python
print(3.0 == 3)        # numbers compare by value
print('3' == 3)        # a string is never equal to a number
````

### Combining conditions: `and`, `or`, `not`
- `and`: **both** must be true
- `or`: **at least one** must be true
- `not`: flips True ↔ False

Truth table for the two-condition operators:

| A | B | A and B | A or B |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

````python
day = "sunday"
print(day == "sunday" or day == "saturday")     # weekend?

is_raining = False
print(not is_raining)
````

### 🆕 More examples

````python
age = 16
has_ticket = True

print(age >= 13 and has_ticket)       # old enough AND has a ticket
print(age >= 18 or has_ticket)        # adult OR has a ticket
print(not (age >= 18))
````

Python lets you **chain** comparisons, just like in maths class:

````python
score = 75
print(60 <= score < 80)        # is score between 60 (included) and 80 (excluded)?
print(0 < score < 50)
````

**Precedence:** `not` is evaluated first, then `and`, then `or`. So `True or True and False` is `True`, because the `and` goes first. Parentheses make your meaning clear:

````python
print(True or True and False)
print((True or True) and False)
````

### Truthy and falsy values
`bool()` converts anything to `True`/`False`. The rule: **zero and "empty" things are `False`; everything else is `True`.**

````python
print(bool(1), bool(-5), bool(0))
print(bool("hello"), bool(""))
print(bool([1, 2]), bool([]))
print(bool(None))
````

## 7. Conditionals: `if`, `elif`, `else`

Conditionals let your program **make decisions**.

```
if condition:
    code that runs when the condition is True
elif other_condition:
    code that runs when the first is False and this is True
else:
    code that runs when nothing above was True
```

Two key things: the **colon `:`** at the end of the line, and the **indentation** (4 spaces) which shows what belongs inside the block. Indentation is not decoration in Python; it is the syntax.

### 📚 From class

````python
age = 10
if age >= 18:
    print("You are an adult.")
print("end of the program")        # not indented: always runs
````

````python
age = int(input("Enter your age: "))
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
````

**`if` vs `elif`.** Separate `if` statements are all checked. With `elif`, Python stops at the **first** true one.

````python
marks = 90
if marks >= 80:
    print("You got a B grade.")
if marks >= 90:
    print("You got an A grade.")        # both print!
````

````python
marks = int(input("Enter your marks: "))
if marks >= 90:
    print("You got an A grade.")
elif marks >= 80:
    print("You got a B grade.")
elif marks >= 70:
    print("You got a C grade.")
else:
    print("You got an F grade.")
````

> ⚠️ **Order matters.** Python checks from the top and stops at the first `True`. If you wrote `marks >= 70` first, a student with 95 would get a C!

### Nested `if`, `pass`, and one-line forms

````python
age = 25
has_license = True

if age >= 18:
    if has_license:
        print("You can drive")
    else:
        print("You need a license")
else:
    print("You are too young to drive")
````

````python
# The same decision, using "and" instead of nesting:
temperature = 25
is_sunny = True
if temperature > 20 and is_sunny:
    print("Perfect beach weather!")
````

`pass` is a placeholder meaning "do nothing yet". An empty block is an error, so `pass` keeps Python happy while you are still planning.

````python
if temperature > 20:
    pass          # TODO: decide what to do on warm days
````

**The ternary (one-line) `if`:** `value_if_true if condition else value_if_false`

````python
a = 10
b = 20
bigger = a if a > b else b
print("Bigger is", bigger)

username = ""
display_name = username if username else "Guest"      # "" is falsy
print("Welcome,", display_name)
````

## 8. `match` / `case`

`match` compares one value against several patterns. It is neat when you have a long chain of `elif x == ...` checks, such as a **menu**. It is similar to `switch` in Java/C#, but there is **no `break`**: only the matching case runs.

`case _:` is the **default** (it matches anything). The underscore is not in quotes!

### 📚 From class

````python
day = 7
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid day")
````

> ⚠️ **Common mistake:** writing `case '_':` (with quotes) means "match the one-character text `_`", not "match anything". Without the quotes, `case _:` is the real default. Test it with `day = 9`.

### 🆕 More examples

`|` means "or" inside a pattern, so several values can share one case. This is also the structure you used for menus in the mini projects.

````python
choice = input("Choose a colour (r/g/b): ")

match choice:
    case "r" | "R":
        print("Red")
    case "g" | "G":
        print("Green")
    case "b" | "B":
        print("Blue")
    case _:
        print("I don't know that colour")
````

````python
day = 9
match day:
    case 1 | 7:
        print("Weekend (in some countries)")
    case 2 | 3 | 4 | 5 | 6:
        print("Weekday")
    case _:
        print("Invalid day")
````

## 9. Loops: `while` and `for`

A loop **repeats** code. Without loops, printing "Hello" 1,000 times would need 1,000 lines.

### The `while` loop
*"Keep repeating **while** a condition is True."* A good `while` loop has three parts:
1. a **starting value**
2. a **condition**
3. an **update** that eventually makes the condition False

If you forget the update you get an **infinite loop** (in Jupyter, stop it with the ■ button).

### 📚 From class

````python
i = 2                  # 1. start
while i < 10:          # 2. condition
    print(i)
    i += 1             # 3. update
````

`break` leaves the loop immediately. `while True:` runs forever until a `break` happens. This is how our **guessing game** worked:

````python
number = 5
while True:
    guess = int(input("Enter a Number: "))
    if number == guess:
        print("Correct")
        break
    elif guess > number:
        print("Too high")
    else:
        print("Too low")
````

### The `for` loop and `range()`
Use `for` when you know **what you are looping over**. `range()` generates numbers:

| Call | Numbers produced |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(1, 10, 2)` | 1, 3, 5, 7, 9 |
| `range(10, 0, -1)` | 10, 9, 8, … 1 |

> 💡 `range(5)` starts at **0** and stops **before** 5. The end is never included.

````python
for i in range(10, 0, -1):
    print(i, end=" ")
print("Liftoff!")
````

````python
for i in range(1, 6):
    print("*" * i)
````

**Accumulator pattern:** start a total at 0 and add to it each time round the loop.

````python
total = 0
for number in range(1, 11):
    total += number
print("The sum of numbers from 1 to 10 is:", total)
````

### Nested loops
A loop inside a loop. For each step of the **outer** loop, the **inner** loop runs completely.

````python
for i in range(3):
    for j in range(3):
        print(i, j)
````

````python
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
    for y in fruits:
        print(x, y)
````

**A grid built with nested loops** (a 4×4 matrix of zeros):

````python
size = 4
matrix = []
for r in range(size):
    row = []
    for c in range(size):
        row.append(0)
    matrix.append(row)

for row in matrix:
    print(row)
````

**An ATM menu** is a `while True` loop with `match` inside, and `break` on exit. This pattern is in every project you build.

````python
balance = 1000

while True:
    print("\n1. Check balance  2. Deposit  3. Withdraw  4. Exit")
    choice = input("Choose: ")

    match choice:
        case "1":
            print("Balance:", balance)
        case "2":
            balance += int(input("Amount to deposit: "))
        case "3":
            amount = int(input("Amount to withdraw: "))
            if amount > balance:
                print("Not enough money!")
            else:
                balance -= amount
        case "4":
            print("Goodbye!")
            break
        case _:
            print("Invalid choice")
````

**`while` or `for`?** If you know how many times (or what to loop over) → `for`. If you repeat *until something happens* → `while`.

| `KeyError` | Dictionary key doesn't exist |

## 11. Modules: `random`, `sys`, `math`

A **module** is a file of ready-made Python code that you can use. Python's **standard library** comes with many. You load one with `import`.

```python
import random                  # use as random.randint(...)
import math as m               # give it a shorter name
from math import sqrt, pi      # import only what you need
```
> ⚠️ Avoid `from module import *`: if two modules have a function with the same name, the second silently replaces the first.

Use `dir(module)` to see what is inside and `help(...)` to read the documentation.

### 📚 From class

````python
import random

print(random.randint(1, 10))          # a whole number from 1 to 10 (both included)

fruits = ["apple", "banana", "cherry"]
print(random.choice(fruits))          # pick one item at random

random.shuffle(fruits)                # shuffle the list in place
print(fruits)
````

````python
import sys
print(sys.version)                    # which Python are you running?
````

**Dice game idea:** use `random` and a loop to roll two dice until you get a double.

````python
import random

rolls = 0
while True:
    d1 = random.randint(1, 6)
    d2 = random.randint(1, 6)
    rolls += 1
    print(f"Roll {rolls}: {d1} and {d2}")
    if d1 == d2:
        print("Double! It took", rolls, "rolls.")
        break
````
