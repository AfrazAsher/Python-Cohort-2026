# 04 · Functions

A **function** is a named block of code that you can run again and again. You have already used many (`print`, `len`, `input`, `range`). Now you will write your own.

Why bother?
- **No repeating yourself:** write once, use many times (the *DRY* rule: Don't Repeat Yourself).
- **Organised code:** big programs are built from small functions with clear names.
- **Easier fixing:** fix a bug in one place.

## 1. Defining and calling functions

```
def function_name(parameters):
    """optional description"""
    code (indented)
    return something
```

- `def` creates ("defines") the function. **Nothing runs yet.**
- To **run** it you **call** it by name with parentheses: `function_name()`.
- Function names follow the same rules as variables: use `snake_case`, make them **verbs** (`print_menu`, `calculate_total`).

### 📚 From class

````python
def greet():
    print("Hello! Welcome to the program.")

greet()
greet()
greet()
````

Defining a function and calling it are two separate steps. A function you only define never runs:

````python
def never_called():
    print("You will never see this")

# (no call here, so nothing is printed)
````

### 🆕 More examples

````python
def print_menu():
    print("\n===== Mini Game =====")
    print("1. Start")
    print("2. Options")
    print("3. Quit")

print_menu()
````

## 2. Parameters and arguments

A function becomes much more useful when it can receive **inputs**.

- **Parameter** = the variable name in the `def` line (`name`)
- **Argument** = the actual value you pass when calling (`"Ahmed"`)

### 📚 From class

````python
def greet(name):
    print(f"Hello, {name}! Welcome to the program.")

greet("Ahmed")
greet("Sara")
````

````python
def add_numbers(a, b):
    print(a + b)

add_numbers(5, 3)
````

The number of arguments must match the parameters:

````python
def add(a, b):
    return a + b

add(1, 2, 3)         # three arguments, but the function wants two
````

### 🆕 More examples

````python
def rectangle_area(width, height):
    print(f"A {width} x {height} rectangle has area {width * height}")

rectangle_area(4, 5)
rectangle_area(10, 2.5)
````

Arguments are matched **by position**: the first value goes to the first parameter, and so on. Order matters!

````python
def introduce(name, age):
    print(f"{name} is {age} years old")

introduce("Ali", 15)
introduce(15, "Ali")       # runs, but nonsense output
````

## 3. `return` vs `print`

This confuses almost everyone at first.

- `print` **shows** a value on the screen. A human can read it, but the program cannot use it.
- `return` **hands a value back** to the code that called the function, so you can store it and use it.

A function with **no `return`** gives back the special value `None` (Python's "nothing").

### 📚 From class

````python
def add_numbers(a, b):
    print(a + b)

result = add_numbers(5, 3)    # prints 8
print(result)                 # None! There was nothing to return
````

````python
def add_numbers(a, b):
    return a + b

result = add_numbers(5, 3)
print(result)                 # 8
print(result * 2)             # 16: we can actually USE the value
````

`return` also **ends the function immediately**. Any code after it is skipped:

````python
def check_age(age):
    if age >= 18:
        return "adult"
    return "minor"          # only reached when the if-condition is False

print(check_age(20))
print(check_age(12))
````

### 🆕 More examples

Functions that **answer a yes/no question** usually return a `bool`. Name them like a question (`is_...`, `has_...`):

````python
def is_even(n):
    return n % 2 == 0

print(is_even(10))
print(is_even(7))

numbers = [1, 2, 3, 4, 5, 6]
print([n for n in numbers if is_even(n)])
````

**Returning more than one value** really returns one tuple (see the Tuples section of the previous notebook):

````python
def stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

low, high, average = stats([4, 8, 15, 16, 23, 42])
print(low, high, average)
````

> ⚠️ **Common mistake:** printing inside a function when you meant to return, then wondering why `result * 2` crashes with `None`. If another part of your program needs the answer → **return it**.

````python
def convert_to_f(celsius):
    return celsius * 9 / 5 + 32

for c in [0, 25, 100]:
    print(f"{c}°C = {convert_to_f(c)}°F")
````

## 4. Default and keyword arguments

### Default arguments
Give a parameter a value in the `def` line and it becomes **optional**. If the caller leaves it out, the default is used.

### 📚 From class

````python
def greet(name="Student"):
    print(f"Hello, {name}!")

greet()             # uses the default
greet("Ahmed")      # uses what we passed
````

````python
def greet(name, age="unknown"):
    print(f"Hello, I'm {name} and my age is {age}")

greet("ali", "18")
greet(age=25, name="ahmed")      # keyword arguments: ORDER doesn't matter
greet("sana")
````

````python
def my_function(animal, name, age):
    print("I have a", age, "year old", animal, "named", name)

my_function("dog", name="Buddy", age=5)     # mixing: positional first, then keyword
````

**Rules**
1. Parameters **with** defaults go **after** those without.
2. When calling, **positional** arguments come first, then **keyword** ones.

### 🆕 More examples

````python
def power(base, exponent=2):
    return base ** exponent

print(power(5))          # 25: squared by default
print(power(2, 10))      # 1024
print(power(exponent=3, base=4))
````

`print` itself has keyword arguments with defaults: `sep=" "` and `end="\n"`. That is why `print(a, b, sep="-")` works.

> ⚠️ **Never use a list or dictionary as a default value** (`def f(items=[])`). The same list is shared between calls. Use `None` instead:

````python
def add_item(item, basket=None):
    if basket is None:
        basket = []          # a fresh list every call
    basket.append(item)
    return basket

print(add_item("apple"))
print(add_item("banana"))
````

## 5. Docstrings and `help()`

A **docstring** is a triple-quoted string on the first line of a function. It describes what the function does. `help(function)` displays it. Good programmers write them.

````python
def least_difference(a, b, c):
    """Return the smallest difference between any two numbers among a, b and c.

    >>> least_difference(1, 5, -5)
    4
    """
    diff1 = abs(a - b)
    diff2 = abs(b - c)
    diff3 = abs(a - c)
    return min(diff1, diff2, diff3)

print(least_difference(1, 10, 100))
help(least_difference)
````

You can use `help()` on built-in functions too. Always pass the function **name** (without `()`), not a call:

````python
help(round)
````

## 6. Scope: local vs global variables

Variables created **inside** a function are **local**: they exist only while the function runs. Variables created outside are **global**.

Think of a function as a room with a closed door: what is created inside stays inside. The room can **look out** of the window (read globals), but if it writes to a name, it creates a **new local** variable.

### 📚 From class

````python
count = 10

def change_count():
    count = 5                       # a NEW local variable, not the global one
    print("Inside function:", count)

change_count()
print("Outside function:", count)   # still 10
````

````python
def show_count():
    print("Global count is:", count)    # READING a global is fine

show_count()
````

### 🆕 More examples

````python
def make_secret():
    secret = 42
    return secret

make_secret()
print(secret)         # secret only existed inside the function
````

To change a global from inside a function, you need the `global` keyword. **But avoid it when you can.** It is better to pass values in as parameters and `return` results.

````python
score = 0

def add_point():
    global score
    score += 1

add_point()
add_point()
print(score)
````

````python
# The cleaner way: no global needed
def add_point_clean(score):
    return score + 1

score = 0
score = add_point_clean(score)
score = add_point_clean(score)
print(score)
````

## 7. `*args` and `**kwargs`

Sometimes you don't know **how many** arguments will be passed.

| Syntax | Collects | Into a | Used for |
|---|---|---|---|
| `*args` | extra **positional** arguments | **tuple** | any number of values |
| `**kwargs` | extra **keyword** arguments | **dictionary** | any number of `name=value` pairs |

The names `args` and `kwargs` are only a convention. The stars are what matter.

### 📚 From class

````python
def add_numbers(a, b):
    return a + b

add_numbers(1, 2, 3, 4)    # too many!
````

````python
def add_numbers(*args):
    print(args)
    print(type(args))
    total = 0
    for number in args:
        total += number
    return total

print(add_numbers(1, 2, 3, 4, 5))
print(add_numbers(10, 20))
print(add_numbers())          # zero arguments is fine
````

````python
def show_profile(**kwargs):
    print(kwargs)
    print(type(kwargs))
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_profile(name="John", age=30, city="New York")
show_profile(name="Sara", favorite_language="Python")
````

You can combine them. The order must be: normal parameters, then `*args`, then `**kwargs`.

````python
def create_student(name, *args, **kwargs):
    print("Name:", name)
    print("Extra info:", args)
    print("Details:", kwargs)

create_student("Ali", "9th grade", school="City School", favorite_subject="Math")
````

### 🆕 More examples

````python
def average(*numbers):
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

print(average(80, 90, 100))
print(average(5))
print(average())
````

The `*` also works in reverse when **calling**: it unpacks a list into separate arguments.

````python
def volume(length, width, height):
    return length * width * height

dimensions = [2, 3, 4]
print(volume(*dimensions))            # same as volume(2, 3, 4)

settings = {"length": 5, "width": 2, "height": 2}
print(volume(**settings))             # same as volume(length=5, width=2, height=2)
````

`print` is itself an `*args` function. That is why you can give it any number of values.

## 8. Lambda functions

A **lambda** is a tiny, **nameless** function written in one line:

```
lambda parameters: expression
```

The expression's result is automatically returned. It is the same as a short `def`.

### 📚 From class

````python
def add(a, b):
    return a + b

add_lambda = lambda a, b: a + b
print(add(5, 5))
print(add_lambda(2, 3))
````

````python
def my_func(n):
    return lambda a: a * n          # a function that returns a function!

doubler = my_func(2)
print(doubler(11))
print(my_func(3)(11))
````

````python
square = lambda x: x ** 2
is_even = lambda n: n % 2 == 0
print(square(5), is_even(10))
````

> 💡 You usually don't store a lambda in a variable. If it needs a name, write a normal `def`. Lambdas shine when you need a **quick throwaway function**, mostly as the `key=` of `sorted`, `min` and `max`. That's the next section.

## 9. Functions that use functions: `key=`, `map`, `filter`

In Python, a function is a **value** like any other: you can store it in a variable or pass it to another function. Functions that take functions are called **higher-order functions**.

### `key=` with `sorted`, `min`, `max`
`key` tells Python **what to look at** when comparing items.

### 📚 From class

````python
students = [("Ali", 85), ("Sara", 92), ("Ahmed", 70)]

by_marks = sorted(students, key=lambda student: student[1])
print(by_marks)

best_first = sorted(students, key=lambda student: student[1], reverse=True)
print(best_first)
````

````python
word_counts = {"python": 5, "code": 2, "loop": 8}
sorted_words = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)
print(sorted_words)
````

### 🆕 More examples

````python
words = ["banana", "kiwi", "apple", "fig"]

print(sorted(words))                          # alphabetical
print(sorted(words, key=len))                 # by length
print(sorted(words, key=lambda w: w[-1]))     # by last letter
print(max(words, key=len))                    # the longest word
print(min(words, key=lambda w: w.lower()))
````

````python
people = [
    {"name": "Ali", "age": 31},
    {"name": "Sara", "age": 24},
    {"name": "Zain", "age": 28},
]
youngest = min(people, key=lambda p: p["age"])
print(youngest)
print([p["name"] for p in sorted(people, key=lambda p: p["age"])])
````

Using a normal named function as the key works too (Kaggle's example): which number is biggest **modulo 5**?

````python
def mod_5(x):
    return x % 5

print(max(100, 51, 14))
print(max(100, 51, 14, key=mod_5))
````

### `map` and `filter` *(bonus)*
- `map(function, items)` → apply the function to **every** item
- `filter(function, items)` → keep only the items where the function returns `True`

They return lazy objects, so wrap them in `list(...)` to see the result. List comprehensions do the same job and are often easier to read.

````python
nums = [1, 2, 3, 4, 5, 6]

print(list(map(lambda n: n * n, nums)))
print(list(filter(lambda n: n % 2 == 0, nums)))

# the comprehension versions:
print([n * n for n in nums])
print([n for n in nums if n % 2 == 0])
````

## 10. Recursion

A **recursive** function is a function that **calls itself**. It is a way to solve a big problem by solving a smaller version of the same problem.

Every recursive function needs two things:
1. **Base case**: when to **stop** (the simplest case, answered directly)
2. **Recursive case**: call itself with a **smaller** problem, moving toward the base case

> ⚠️ Without a base case, a function calls itself forever, until Python stops it with a `RecursionError`.

### 📚 From class

````python
def countdown(n):
    if n <= 0:                      # base case: stop here
        print("Liftoff! 🚀")
        return
    print(n)
    countdown(n - 1)                # recursive case: a smaller problem

countdown(3)
````

**How `countdown(3)` really runs.** Each call waits for the next one to finish:

```
countdown(3) starts
   prints 3
   calls countdown(2)
        countdown(2) starts
           prints 2
           calls countdown(1)
                countdown(1) starts
                   prints 1
                   calls countdown(0)
                        countdown(0): n <= 0, prints "Liftoff!" and returns
                countdown(1) finishes
        countdown(2) finishes
countdown(3) finishes
```

**Factorial:** `5! = 5 × 4 × 3 × 2 × 1 = 120`. Notice `5! = 5 × 4!`: the answer is built from a smaller factorial.

````python
def factorial(x):
    if x == 0:                      # base case (0! is 1)
        return 1
    return x * factorial(x - 1)

print(factorial(5))
````

```
factorial(5) = 5 * factorial(4)
factorial(4) = 4 * factorial(3)
factorial(3) = 3 * factorial(2)
factorial(2) = 2 * factorial(1)
factorial(1) = 1 * factorial(0)
factorial(0) = 1                    ← base case, no more recursion

Now it unwinds back up:
factorial(1) = 1 * 1  = 1
factorial(2) = 2 * 1  = 2
factorial(3) = 3 * 2  = 6
factorial(4) = 4 * 6  = 24
factorial(5) = 5 * 24 = 120
```

**Fibonacci:** each number is the sum of the previous two: `0, 1, 1, 2, 3, 5, 8, 13…`

````python
def fibonacci(n):
    if n <= 1:                       # base cases: fib(0)=0, fib(1)=1
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(5))
print([fibonacci(i) for i in range(10)])
````

> 💡 This version is **slow** for big `n` because it recalculates the same values over and over (`fibonacci(40)` takes noticeable time). It is a good lesson in recursion, but not the way to do it in real life.

### 🆕 More examples

**Sum of a list:** the sum is the first item plus the sum of the *rest*.

````python
def list_sum(numbers):
    if len(numbers) == 0:            # base case: empty list sums to 0
        return 0
    return numbers[0] + list_sum(numbers[1:])

print(list_sum([1, 2, 3, 4, 5]))
````

The same thing with a loop. **Any recursive function can be written with a loop**, and the loop version is usually faster and uses less memory. Recursion is nicest when the problem is naturally nested (folders inside folders, family trees, puzzles).

````python
def list_sum_loop(numbers):
    total = 0
    for num in numbers:
        total += num
    return total

print(list_sum_loop([1, 2, 3, 4, 5]))
````

**Reversing a string** recursively: the reverse is the *last* letter plus the reverse of everything before it.

````python
def reverse(text):
    if len(text) <= 1:
        return text
    return text[-1] + reverse(text[:-1])

print(reverse("python"))
````

**What happens with no base case?** Python protects itself with a limit (about 1,000 nested calls):

````python
def broken_countdown(n):
    broken_countdown(n - 1)         # no base case: never stops!

try:
    broken_countdown(5)
except RecursionError as e:
    print("RecursionError:", e)
````

## 11. Putting it together: a menu-driven mini program

Here is the pattern from our projects. Every action is a **small function**; the main loop just chooses which to call. Notice how readable the loop becomes.

````python
def print_menu():
    print("\n===== To-Do List =====")
    print("1. Add a task")
    print("2. Show tasks")
    print("3. Complete a task")
    print("4. Exit")

def add_task(tasks, text):
    tasks.append(text)
    print(f"Added: {text}")

def show_tasks(tasks):
    if not tasks:
        print("Nothing to do!")
        return
    for number, task in enumerate(tasks, start=1):
        print(f"{number}. {task}")

def complete_task(tasks, number):
    if 1 <= number <= len(tasks):
        done = tasks.pop(number - 1)
        print(f"Completed: {done}")
    else:
        print("No such task.")

tasks = []
while True:
    print_menu()
    choice = input("Choose: ")
    match choice:
        case "1":
            add_task(tasks, input("Task: "))
        case "2":
            show_tasks(tasks)
        case "3":
            complete_task(tasks, int(input("Task number: ")))
        case "4":
            print("Bye!")
            break
        case _:
            print("Invalid choice")
````

> ⚠️ **Don't name your variables or functions after built-ins.** In class we accidentally wrote `def sum(a, b)`. That *replaced* Python's `sum`, so a later `sum([1, 2, 3])` failed! Use names like `add` or `total`.

### Common mistakes checklist
- Forgetting the **colon** after `def ...():`
- Forgetting to **indent** the function body
- Defining a function but **never calling** it
- Using `print` where you need `return`
- Forgetting the **base case** in recursion
- Modifying a global without `global` (or, better, passing it in)
