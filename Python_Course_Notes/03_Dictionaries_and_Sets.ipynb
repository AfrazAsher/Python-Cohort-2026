## 1. Dictionaries: key → value

Think of a real dictionary: you look up a **word** (the *key*) and get its **meaning** (the *value*). A Python dict stores **key-value pairs**:

```python
contact = {"name": "afraz", "phone": "0322-4884403", "age": 26}
#            key    value
```
- Keys are usually strings or numbers and must be **unique** and **unchangeable** (a tuple can be a key, a list can't).
- Values can be **anything**: numbers, strings, lists, even other dictionaries.
- Lookup by key is **very fast**, even with a million items.

> 💡 **Fun fact:** the same idea has other names: *HashMap* in Java, *Dictionary<K,V>* in C#, *object* in JavaScript, *associative array* in PHP.

### 📚 From class

````python
contact = {"name": "afraz", "phone": "0322-4884403", "age": 26}
print(contact)
print(type(contact))
print(contact["name"])
````

Compare with a list. With a list you must **remember** that index 1 is the phone number. With a dictionary the data **describes itself**:

````python
contact_list = ["afraz", "0322-4884403", 26]
print(contact_list[1])               # what is index 1 again?

print(contact["phone"])               # no guessing
````

### 🆕 More examples

````python
student = {
    "name": "Sara",
    "grade": 9,
    "subjects": ["Math", "Physics"],      # a value can be a list
    "address": {"city": "Lahore", "zip": "54000"},   # or another dict
}
print(student["subjects"][0])
print(student["address"]["city"])
print(len(student))                        # number of key-value pairs
````

## 2. Reading, adding, changing, deleting

| Task | Code |
|---|---|
| Read | `d[key]` (error if missing) or `d.get(key, default)` (safe) |
| Add **or** change | `d[key] = value` |
| Delete | `del d[key]` or `d.pop(key)` |
| Is the key there? | `key in d` |
| Merge another dict | `d.update(other)` |
| Remove everything | `d.clear()` |

> ⚠️ `d[key]` raises a `KeyError` if the key doesn't exist. `d.get(key, default)` never crashes.

### 📚 From class

````python
contact = {"name": "afraz", "phone": "0322-4884403", "age": 26}

print(contact.get("bmi", "not found"))    # safe lookup with a default

contact["bmi"] = 25                       # adding a new key
print(contact["bmi"])

del contact["bmi"]                        # delete a key
print(contact.keys())
print(contact.values())
print(contact.items())
````

````python
print(contact["bmi"])        # the key is gone, so this crashes
````

### 🆕 More examples

````python
prices = {"apple": 120, "banana": 60}

prices["cherry"] = 300          # new key
prices["apple"] = 140           # existing key: value is REPLACED
print(prices)

print("banana" in prices)       # checks KEYS
print("mango" in prices)
print(120 in prices)            # False: 120 is a value, not a key

removed = prices.pop("banana")  # remove AND get the value back
print("Removed price:", removed)
print(prices)
````

````python
defaults = {"theme": "light", "font": 12}
user = {"font": 16}
defaults.update(user)           # user settings override defaults
print(defaults)
````

## 3. Looping over a dictionary

- `for k in d:` → each **key**
- `d.keys()` → all keys, `d.values()` → all values
- `d.items()` → **(key, value) pairs**, usually the most useful. Because each item is a tuple, we unpack it with `for key, value in d.items()`.

### 📚 From class

````python
contact = {"name": "afraz", "phone": "0322-4884403", "age": 26}
for x, y in contact.items():
    print(x, y)
````

### 🆕 More examples

````python
prices = {"apple": 140, "cherry": 300, "mango": 250}

for fruit in prices:                       # keys only
    print(fruit)

for price in prices.values():              # values only
    print(price)

total = sum(prices.values())
print("Total:", total)

for fruit, price in prices.items():
    print(f"{fruit:<8} Rs {price}")
````

Finding the **most expensive** item with `max` and a `key` function (more about `key=` in the Functions notebook):

````python
priciest = max(prices, key=prices.get)
print(priciest, prices[priciest])
````

Sorting a dictionary by value turns it into a list of tuples:

````python
word_counts = {"python": 5, "code": 2, "loop": 8}
ranked = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)
print(ranked)
````

## 4. Nested data: lists of dictionaries

Real data usually looks like this: a **list of dictionaries**, where each dictionary is one record (one book, one student, one product). This is the shape of what you get from files and APIs.

Remember our early Library project? Each book was a dictionary inside a list.

### 📚 From class

````python
books = [
    {"title": "Dune", "author": "Frank Herbert", "available": True},
    {"title": "1984", "author": "George Orwell", "available": False},
]

for book in books:
    status = "Available" if book["available"] else "Borrowed"
    print(f"{book['title']} by {book['author']} - {status}")
````

````python
# search for a book
title = "1984"
for book in books:
    if book["title"] == title:
        print("Found:", book)
        break
else:
    print("Not found")
````

### 🆕 More examples

````python
students = [
    {"name": "Ali",  "marks": 85},
    {"name": "Sara", "marks": 92},
    {"name": "Zain", "marks": 70},
]

average = sum(s["marks"] for s in students) / len(students)
print(f"Class average: {average:.1f}")

top = max(students, key=lambda s: s["marks"])
print("Top student:", top["name"])

passed = [s["name"] for s in students if s["marks"] >= 75]
print("Scored 75+:", passed)
````

> 💡 Notice we wrote `book['title']` with **single** quotes inside an f-string that uses **double** quotes. Using the same quote type inside would end the string early (in older Python versions).

## 5. Common dictionary patterns

### Counting things
This is the single most useful dictionary pattern: **counting how many times each item appears.**

````python
text = "the cat and the hat and the bat"
counts = {}

for word in text.split():
    if word in counts:
        counts[word] += 1
    else:
        counts[word] = 1

print(counts)
````

`get` makes it shorter: *"take the current count (or 0 if new) and add 1"*:

````python
counts = {}
for word in text.split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
````

There is even a ready-made tool for counting, in the standard library:

````python
from collections import Counter
print(Counter(text.split()))
print(Counter("mississippi").most_common(2))
````

### Grouping
Group words by their first letter: each key gets a **list**:

````python
words = ["apple", "avocado", "banana", "blueberry", "cherry"]
groups = {}

for w in words:
    letter = w[0]
    if letter not in groups:
        groups[letter] = []        # first time we see this letter: make an empty list
    groups[letter].append(w)

print(groups)
````

### A lookup table
Use a dictionary instead of a long `if/elif` chain:

````python
capitals = {"Pakistan": "Islamabad", "France": "Paris", "Japan": "Tokyo"}

country = input("Which country? ")
print(capitals.get(country, "I don't know that one yet"))
````

## 6. Dictionary comprehensions *(bonus)*

Just like list comprehensions, but with `key: value`:

```
{ key_expression : value_expression   for item in something   if condition }
```

````python
planets = ["Mercury", "Venus", "Earth", "Mars"]
initials = {p: p[0] for p in planets}
print(initials)
````

````python
squares = {n: n ** 2 for n in range(1, 6)}
print(squares)

long_words = {w: len(w) for w in ["python", "is", "awesome"] if len(w) > 2}
print(long_words)
````

Swapping keys and values:

````python
capitals = {"Pakistan": "Islamabad", "France": "Paris"}
reverse = {city: country for country, city in capitals.items()}
print(reverse)
````

## 7. Sets

A **set** is an unordered collection of **unique** items. Think of a bag of stickers where duplicates simply disappear.

- Duplicates are removed **automatically**.
- No order and **no indexing**: `my_set[0]` is an error.
- Checking `x in my_set` is extremely fast.

> ⚠️ **Gotcha:** `{}` is an **empty dictionary**, not an empty set! Use `set()` for an empty set.

### 📚 From class

````python
fruits = ["apple", "banana", "cherry", "apple"]       # a LIST: keeps the duplicate
fruits_set = set(fruits)                               # a SET: duplicate gone
print(fruits)
print(fruits_set)
````

> 💡 A set has **no order**, so the items may be printed in a different order on your computer. That is normal and not a mistake.

````python
empty = {}
print(type(empty))               # dict! NOT a set

actual_empty_set = set()
print(type(actual_empty_set))
````

### Set methods

| Method | What it does |
|---|---|
| `s.add(x)` | add **one** item |
| `s.remove(x)` | remove (error if missing) |
| `s.discard(x)` | remove (**no** error if missing) |
| `s.pop()` | remove a **random** item (sets have no first/last) |
| `s.clear()` | empty it |
| `x in s` | membership test |

````python
fruits_set.add("orange")
print(fruits_set)

fruits_set.discard("berry")      # safe: "berry" isn't there, no error
fruits_set.remove("orange")      # fine, it is there
print(fruits_set)
````

````python
fruits_set.remove("berry")        # "berry" isn't there: KeyError
````

````python
print(fruits_set[0])              # no indexing in a set
````

### 🆕 More examples

The most common use of a set: **remove duplicates** and **count unique things**.

````python
numbers = [1, 2, 2, 3, 3, 3, 4]
unique_numbers = set(numbers)
print(unique_numbers)
print("Unique count:", len(unique_numbers))

text = "to be or not to be"
words = text.split()
unique_words = set(words)
print(f"You used {len(unique_words)} unique words out of {len(words)} total.")
````

Want to **keep the original order** while removing duplicates? A set alone won't do (no order). Use a dictionary trick, since dicts remember insertion order:

````python
print(list(dict.fromkeys([3, 1, 3, 2, 1])))
````

A set **comprehension** works the same way (bonus):

````python
print({len(w) for w in ["hi", "hello", "hey", "howdy"]})
````

## 8. Set operations

Sets are borrowed from maths class (Venn diagrams!). Two groups of students:

| Operation | Operator | Method | Meaning |
|---|---|---|---|
| **Union** | `a \| b` | `a.union(b)` | in either |
| **Intersection** | `a & b` | `a.intersection(b)` | in **both** |
| **Difference** | `a - b` | `a.difference(b)` | in `a` but **not** `b` |
| **Symmetric difference** | `a ^ b` | `a.symmetric_difference(b)` | in **exactly one** |

### 📚 From class

````python
math_students = {"ali", "ahmed", "sana", "Adam"}
science_students = {"ali", "ahmed", "sana", "john"}

print("union of both classes:", math_students | science_students)
print("intersection of both classes:", math_students & science_students)
print("difference of both classes:", math_students - science_students)
print("symmetric difference:", math_students ^ science_students)
````

The class activity: find the **common words** in two sentences.

````python
sen_one = input("Enter sentence one: ").lower().split()
sen_two = input("Enter sentence two: ").lower().split()

print("Common words:", set(sen_one) & set(sen_two))
````

### 🆕 More examples

````python
class_a = {"Sara", "Ahmed", "Zain", "Bilal"}
class_b = {"Sara", "Ahmed", "Hira"}

print("In both classes:", class_a & class_b)
print("Everyone:", class_a | class_b)
print("Only in A:", class_a - class_b)
print("Only in B:", class_b - class_a)
print("In exactly one class:", class_a ^ class_b)
````

Subset tests:

````python
required = {"python", "git"}
skills = {"python", "git", "sql"}
print(required.issubset(skills))          # does she have all the required skills?
print(skills.issuperset(required))
print({"java"}.isdisjoint(skills))        # nothing in common?
````

**Real use:** which tags do two posts share?

````python
post1 = {"python", "beginner", "tutorial"}
post2 = {"python", "advanced", "tutorial"}
print("Shared tags:", sorted(post1 & post2))
````

## 9. Which structure should I use?

| I need to… | Use |
|---|---|
| keep items **in order**, allow duplicates, change them | **list** |
| store a **fixed** group of values | **tuple** |
| look up a value by a **name/label** | **dictionary** |
| store **unique** items / remove duplicates / test membership quickly | **set** |
| keep many **records**, each with several fields | **list of dictionaries** |

**Mutable or immutable?**

| Mutable (can change) | Immutable (can't change) |
|---|---|
| list, dict, set | int, float, str, bool, tuple |
