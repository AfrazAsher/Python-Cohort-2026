## 1. Why classes? Classes and objects

Imagine a library with 500 books. With dictionaries you could write `{"title": ..., "author": ...}` 500 times, and every function that uses a book must *know* the keys. Easy to make a typo, and the data and the actions are stored in different places.

A **class** is a **blueprint** (like a cookie cutter). An **object** (also called an **instance**) is a real thing made from that blueprint (a cookie).

- The class `Car` describes what every car has (brand, speed) and can do (drive, brake).
- The objects `my_car` and `your_car` are two separate cars made from that blueprint, each with its own values.

> 💡 **Real life:** a house **blueprint** is a class. Each actual house built from it is an object. You can build many houses (objects) from one blueprint (class), and painting one house red doesn't repaint the others.

| Word | Meaning | Example |
|---|---|---|
| **class** | the blueprint | `Dog` |
| **object / instance** | one thing made from it | `my_dog = Dog("Rex")` |
| **attribute** | data stored in an object | `my_dog.name` |
| **method** | a function that belongs to the class | `my_dog.bark()` |

Here is the smallest class possible. It is empty (`pass` means "nothing here yet"), but it already works:

````python
class Course:
    pass

course1 = Course()                 # create an object (instantiate)
course1.name = "Mathematics"       # attach data to it afterwards
course1.credits = 3
print(f"Course Name: {course1.name}, Credits: {course1.credits}")
````

It works, but it's **not the ideal way**: every object needs manual setup, it is easy to forget an attribute, and typos create new attributes silently. The solution is `__init__`.

## 2. `__init__` and `self`

`__init__` is the **constructor**. Python runs it **automatically** whenever you create an object. Its job is to set up the object's starting data.

**`self`** is the object itself: the specific one that is being created or used right now. It is always the first parameter of a method, and Python passes it in for you (you never type it when calling). It is the same as `this` in Java and C#.

```
class Student:
    def __init__(self, name, age=18):
        self.name = name      # store the value INSIDE the object
        self.age = age
```

`self.name = name` means: *"this object's `name` attribute = the value that was passed in."*

### 📚 From class

````python
class Student:
    city = "Lahore"                 # a class variable shared by all instances (more below)

    def __init__(self, name, age=18):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}, City: {self.city}")


student1 = Student("Alice", 20)
student1.display_info()

student2 = Student("Bob", 22)
student2.display_info()

student3 = Student("Charlie")        # age defaults to 18
student3.display_info()
````

Each object keeps **its own copy** of the data. Changing one doesn't touch the others:

````python
student1.age = 21
student1.display_info()

student2.name = "Robert"
student2.display_info()
student3.display_info()              # untouched
````

You can even delete an attribute from one object (rarely useful, but it shows attributes are just names stored in the object):

````python
del student1.age
student1.display_info()              # the method still expects self.age
````

### 🆕 More examples

````python
class Dog:
    def __init__(self, name, breed, age):
        self.name = name
        self.breed = breed
        self.age = age

rex = Dog("Rex", "German Shepherd", 3)
bella = Dog("Bella", "Labrador", 5)

print(rex.name, rex.breed, rex.age)
print(bella.name, bella.breed, bella.age)
print(rex is bella)             # two different objects
````

> ⚠️ **Common mistakes**
> - Forgetting `self` as the first parameter → `TypeError: ... takes 1 positional argument but 2 were given`
> - Writing `name = name` instead of `self.name = name` → the data is thrown away at the end of `__init__`
> - Using the name `__init__` with the wrong underscores (it needs **two** on each side)

````python
class Broken:
    def __init__(self, name):
        name = name              # forgot self.: nothing is stored on the object

b = Broken("test")
print(b.name)
````

## 3. Methods

A **method** is a function that lives inside a class. It always takes `self` first, which gives it access to the object's attributes (`self.age`) and to its other methods (`self.greet()`).

### 📚 From class

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        return f"Hello {self.name}"

    def welcome(self):
        message = self.greet()                  # a method calling another method
        print(f"Welcome to the World, {message}")

    def celebrate_birthday(self):
        self.age += 1                           # a method changing the object's own data
        print(f"Happy Birthday, You are now {self.age}")


p1 = Person("Ali", 25)
p1.welcome()
p1.celebrate_birthday()
p1.celebrate_birthday()
print(p1.age)
````

**Methods can keep track of a changing list** inside the object. Here every student object collects its own grades:

````python
class Student:
    def __init__(self, name):
        self.name = name
        self.grade = []                  # every student gets a NEW empty list

    def add_grade(self, grade):
        self.grade.append(grade)
        print(f"Grade {grade} added for {self.name}")

    def average_grade(self):
        if self.grade:
            avg = sum(self.grade) / len(self.grade)
            print(f"Average grade for {self.name} is {avg}")
        else:
            print(f"No grades available for {self.name}")


student1 = Student("Alice")
student1.add_grade(85)
student1.add_grade(90)
student1.average_grade()

Student("Zain").average_grade()          # a brand new student, no grades
````

### 🆕 More examples

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Not enough money!")
        else:
            self.balance -= amount

acc = BankAccount("Sara", 1000)
acc.deposit(500)
acc.withdraw(200)
acc.withdraw(5000)
print(f"{acc.owner} has Rs {acc.balance}")
````

> 💡 **Method vs function:** `len(my_list)` is a *function*, `my_list.append(5)` is a *method*. You have been using methods since the Lists notebook. Now you can write your own.

## 4. `__str__`: printing objects nicely

What happens when you `print` an object? Let's see:

````python
print(p1)
````

Not very friendly: just the class name and a memory address (yours will show different digits). If you define a special method called **`__str__`**, Python uses it whenever the object is printed or converted with `str()`. It must **return** a string. (It is Python's `toString()`.)

Methods with double underscores on both sides are called **dunder methods** ("double underscore") or *magic methods*. You never call them directly; Python calls them for you.

### 📚 From class

````python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def __str__(self):
        return f"{self.year} {self.brand} {self.model}"


car1 = Car("Toyota", "Camry", 2020)
print(car1)
print(f"I drive a {car1}")
print(str(car1))
````

### 🆕 More examples

````python
class Pizza:
    def __init__(self, flavor, size_cm):
        self.flavor = flavor
        self.size_cm = size_cm

    def __str__(self):
        return f"{self.flavor} pizza ({self.size_cm} cm)"

menu = [Pizza("Margherita", 30), Pizza("Pepperoni", 35), Pizza("BBQ Chicken", 40)]
for pizza in menu:
    print(pizza)            # print calls __str__ for every object
````

> ⚠️ `__str__` must **return** a string, not print one. A `print` inside it shows the text but then returns `None`, which causes an error.

## 5. Class attributes vs instance attributes

| | Instance attribute | Class attribute |
|---|---|---|
| Defined | inside `__init__`: `self.x = ...` | in the class body, **outside** any method |
| Belongs to | **each object** separately | the **class**, shared by every object |
| Example | a student's `name` | the `city` all students share; a counter of how many were created |

### 📚 From class

````python
class Book:
    total_books = 0                         # class attribute: shared by every Book

    def __init__(self, title, author, available=True):
        self.title = title                  # instance attributes: one set per Book
        self.author = author
        self.available = available
        Book.total_books += 1               # every new Book increases the shared count

    def __str__(self):
        status = "Available" if self.available else "Borrowed"
        return f"'{self.title}' by {self.author} - {status}"

    def borrow(self):
        if self.available:
            self.available = False
            print(f"You borrowed '{self.title}'.")
        else:
            print(f"Sorry, '{self.title}' is already borrowed.")

    def return_book(self):
        if not self.available:
            self.available = True
            print(f"Thanks for returning '{self.title}'.")
        else:
            print(f"'{self.title}' wasn't borrowed.")


book1 = Book("1984", "George Orwell")
print(book1)
book1.borrow()
print(book1)
book1.borrow()
book1.return_book()
print(book1)

book2 = Book("Dune", "Frank Herbert")
print("Books created so far:", Book.total_books)
````

### The classic trap
Why do we write `Book.total_books += 1` and not `self.total_books += 1`?

`self.total_books += 1` would create a **brand-new instance attribute** on that one object, hiding the shared one. The shared counter would never change! Always use the **class name** to change a class attribute.

````python
class Counter:
    count = 0

    def __init__(self):
        self.count += 1          # BUG: creates an instance attribute named count

a = Counter()
b = Counter()
c = Counter()
print("Counter.count     :", Counter.count)    # still 0!
print("a.count, b.count  :", a.count, b.count)
````

````python
class Counter:
    count = 0

    def __init__(self):
        Counter.count += 1       # correct: change the shared one

a = Counter()
b = Counter()
c = Counter()
print("Counter.count:", Counter.count)
````

### 🆕 More examples

````python
class Employee:
    company = "TechCorp"             # shared constant-like value
    employee_count = 0

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        Employee.employee_count += 1

e1 = Employee("Ali", 50000)
e2 = Employee("Sara", 65000)

print(e1.company, e2.company)
print("Employees:", Employee.employee_count)

Employee.company = "NewTech"         # change it for EVERYONE
print(e1.company, e2.company)
````

## 6. Composition: objects that contain objects

**Inheritance** (next notebook) means *"is a"*. **Composition** means *"has a"*:

- A **car** *has an* engine.
- A **library** *has* books.
- A **playlist** *has* songs.

An attribute doesn't have to be a number or a string. It can be a **list of objects of another class**. The big idea: the container object (`Library`) **delegates** work to the objects inside it (`Book`) instead of repeating their logic.

> 💡 **Real life:** when you press the gas pedal, the car doesn't know how an engine works. It just tells its engine object to rev up. The engine knows how.

### 📚 From class

````python
class Library:
    def __init__(self):
        self.books = []                      # a list that will hold Book OBJECTS

    def add_book(self, book):
        self.books.append(book)
        print(f"Added '{book.title}' to the library.")

    def view_all(self):
        if not self.books:
            print("No books in the library.")
            return
        print("\n--- Library Collection ---")
        for book in self.books:
            print(book)                       # uses Book.__str__ automatically

    def search(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                return book
        return None

    def borrow_book(self, title):
        book = self.search(title)
        if book:
            book.borrow()                     # delegate to the Book object
        else:
            print(f"'{title}' not found in the library.")

    def return_book(self, title):
        book = self.search(title)
        if book:
            book.return_book()
        else:
            print(f"'{title}' not found in the library.")


Book.total_books = 0                    # reset the shared counter for this demo
library = Library()
library.add_book(Book("1984", "George Orwell"))
library.add_book(Book("Dune", "Frank Herbert"))
library.add_book(Book("The Hobbit", "J.R.R. Tolkien"))

library.view_all()

print("\n--- Borrowing ---")
library.borrow_book("1984")
library.borrow_book("1984")          # already borrowed
library.borrow_book("Harry Potter")  # doesn't exist

print("\n--- Returning ---")
library.return_book("1984")
library.view_all()
print(f"\nTotal books created: {Book.total_books}")
````

Look closely at `borrow_book`. The `Library` does **not** change `book.available` itself. It finds the right `Book` and says *"you handle it"*. The borrowing rules live **in one place** (inside `Book`).

Also notice `search()` **returns** the book (or `None`) instead of printing. Its only job is to *find*. What to do when nothing is found is up to whoever called it.

### 🆕 More examples

````python
class Song:
    def __init__(self, title, artist, seconds):
        self.title = title
        self.artist = artist
        self.seconds = seconds

    def __str__(self):
        minutes, secs = divmod(self.seconds, 60)
        return f"{self.title} - {self.artist} ({minutes}:{secs:02})"


class Playlist:
    def __init__(self, name):
        self.name = name
        self.songs = []

    def add(self, song):
        self.songs.append(song)

    def total_time(self):
        return sum(song.seconds for song in self.songs)

    def show(self):
        print(f"== {self.name} ==")
        for number, song in enumerate(self.songs, start=1):
            print(f"{number}. {song}")
        minutes, secs = divmod(self.total_time(), 60)
        print(f"Total: {minutes}:{secs:02}")


mix = Playlist("Road Trip")
mix.add(Song("Dil Dil Pakistan", "Vital Signs", 258))
mix.add(Song("Imagine", "John Lennon", 183))
mix.add(Song("Yellow", "Coldplay", 269))
mix.show()
````

Another composition: a car that *has an* engine and delegates to it.

````python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower
        self.running = False

    def start(self):
        self.running = True
        return "Engine started"


class MyCar:
    def __init__(self, model, horsepower):
        self.model = model
        self.engine = Engine(horsepower)       # the car CREATES and OWNS an Engine

    def start(self):
        return f"{self.model}: {self.engine.start()} ({self.engine.horsepower} hp)"


print(MyCar("Civic", 158).start())
````

## 7. Encapsulation: protecting your data

Right now anyone can do this:

````python
class Pizza:
    def __init__(self, flavor, size_cm):
        self.flavor = flavor
        self.size_cm = size_cm

pizza = Pizza("Pepperoni", 30)
pizza.size_cm = -500          # nothing stops this!
print(pizza.size_cm)
````

A pizza of **-500 cm** makes no sense. **Encapsulation** means *bundling data with the code that works on it, and controlling access to the data* so it can't be put into an invalid state.

### The underscore conventions
Python has no real `private` keyword like Java/C#. Instead it uses **naming conventions**, a polite signal rather than a lock:

| Name | Meaning | Python enforces it? |
|---|---|---|
| `title` | **public**: use freely | – |
| `_author` | **protected**: "internal, please don't touch from outside" | ❌ just a signal |
| `__copies` | **private**: name gets scrambled ("name mangling") | ⚠️ partly: harder to reach, but possible |

### 📚 From class

````python
class Book:
    def __init__(self, title, author, copies):
        self.title = title          # public
        self._author = author       # "protected": a signal, not a lock
        self.__copies = copies      # "private": scrambled, harder to reach


book1 = Book("Dune", "Ali", 3)
print(book1.title)             # works fine
print(book1._author)           # works: the underscore is only a hint
````

````python
print(book1.__copies)          # AttributeError: the real name was scrambled
````

````python
print(book1._Book__copies)     # name mangling: _ClassName__attribute  -> still reachable!
book1.__copies = -500          # this just creates a NEW attribute with a weird name...
print(book1.__copies)          # ...so now there are two different "copies"
print(book1._Book__copies)     # the real one is still 3
````

So Python doesn't *forbid* access: it relies on programmers being sensible ("we are all consenting adults"). The real protection comes from **getter and setter methods** that check values.

### Getters and setters
- a **getter** returns the value
- a **setter** validates the new value *before* storing it

````python
class Book:
    def __init__(self, title, copies):
        self.title = title
        self._copies = copies

    def get_copies(self):
        return self._copies

    def set_copies(self, value):
        self._copies = value          # no checking yet


book1 = Book("Dune", 3)
print(book1.get_copies())
book1.set_copies(5)
print(book1.get_copies())
````

Now add the **validation**: the setter refuses nonsense. We even call the setter from `__init__`, so a bad starting value is caught too.

````python
class Pizza:
    def __init__(self, flavor, size_cm):
        self.flavor = flavor
        self.set_size_cm(size_cm)         # even the constructor goes through the checks

    def get_size_cm(self):
        return self._size_cm

    def set_size_cm(self, value):
        if not 15 <= value <= 50:
            raise ValueError("Pizza size must be between 15 and 50 cm.")
        self._size_cm = value


pizza = Pizza("Pepperoni", 30)
pizza.set_size_cm(40)
print(pizza.get_size_cm())
````

### 🆕 More examples

````python
class Thermostat:
    def __init__(self, target=20):
        self._target = 20
        self.set_target(target)

    def get_target(self):
        return self._target

    def set_target(self, degrees):
        if 10 <= degrees <= 30:
            self._target = degrees
            return True
        print(f"{degrees}°C is outside the safe range (10 to 30). Keeping {self._target}°C.")
        return False


t = Thermostat()
t.set_target(24)
t.set_target(90)
print("Target is", t.get_target())
````

Hiding **how** something is stored lets you change it later without breaking other code. The `Wallet` below stores cents internally, but the outside world only sees rupees:

````python
class Wallet:
    def __init__(self):
        self._cents = 0

    def add_money(self, rupees):
        if rupees <= 0:
            print("Amount must be positive.")
            return
        self._cents += int(rupees * 100)

    def balance(self):
        return self._cents / 100


w = Wallet()
w.add_money(12.5)
w.add_money(-3)
print(w.balance())
````

> 💡 **Python has a neater way to write getters and setters** called `@property`, which lets `pizza.size_cm = 40` run the setter automatically. You will learn it in the next notebook.
