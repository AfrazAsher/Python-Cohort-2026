# 02 · Lists, Strings & Tuples

| | Brackets | Can you change it? | Holds |
|---|---|---|---|
| **list** | `[1, 2, 3]` | ✅ yes (*mutable*) | anything |
| **string** | `"abc"` | ❌ no (*immutable*) | characters only |
| **tuple** | `(1, 2, 3)` | ❌ no (*immutable*) | anything |

## 1. Lists: creating, indexing, slicing

A **list** stores an ordered group of values in square brackets. The items can be of **different types**, and a list can even contain other lists.

### 📚 From class

````python
names = ["ali", "ahmed", "sana", 7, 10, 20, True]     # mixed types are allowed
for name in names:
    print(name)
````

### Indexing
Each item has a position called an **index**. Counting starts at **0**. Negative indices count from the **end** (`-1` is the last item).

```
index:      0         1        2         3
          "Mercury" "Venus"  "Earth"   "Mars"
index:      -4        -3       -2        -1
```

````python
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune"]

print(planets[0])        # first
print(planets[1])
print(planets[-1])       # last
print(planets[-2])       # second from last
print(len(planets))      # how many items
````

### Slicing
`list[start:stop]` gives a **new list** from `start` up to (but **not including**) `stop`. Leave out `start` for "from the beginning" and `stop` for "to the end". A third number is the **step**.

````python
print(planets[0:3])      # first three
print(planets[:3])       # same thing
print(planets[3:])       # from index 3 to the end
print(planets[1:-1])     # everything except first and last
print(planets[-3:])      # the last three
print(planets[::2])      # every second planet
print(planets[::-1])     # reversed!
````

### 🆕 More examples

A list of lists is a simple **table** (rows and columns). Use two indices: `table[row][column]`.

````python
scores = [
    ["Ali",  85, 90],
    ["Sara", 92, 88],
    ["Zain", 70, 65],
]
print(scores[1])           # a whole row
print(scores[1][0])        # row 1, column 0
print(scores[2][-1])       # last score for Zain
````

Useful built-in functions that work on lists:

````python
primes = [2, 3, 5, 7, 11]
print(len(primes))
print(sum(primes))
print(min(primes), max(primes))
print(sorted([5, 2, 9, 1]))          # returns a NEW sorted list
print("Earth" in planets)             # is it in the list?
print("Pluto" in planets)
````

> ⚠️ **Common mistake:** asking for an index that doesn't exist gives `IndexError`. A list of 8 items has indexes 0 to 7.

````python
print(planets[8])
````

## 2. Changing lists: methods

Lists are **mutable**: you can change them *in place*. A **method** is a function that belongs to an object and is called with a dot: `my_list.append(5)`.

| What you want | Method / statement | Notes |
|---|---|---|
| Add to the end | `lst.append(x)` | adds **one** item |
| Add at a position | `lst.insert(i, x)` | |
| Add many items | `lst.extend(other)` | |
| Remove by value | `lst.remove(x)` | first match only; error if missing |
| Remove by position | `lst.pop(i)` or `del lst[i]` | `pop()` with no argument removes the last, and returns it |
| Remove everything | `lst.clear()` | |
| Sort | `lst.sort()` | changes the list; `reverse=True` for descending |
| Reverse | `lst.reverse()` | |
| Position of a value | `lst.index(x)` | error if missing |
| Count a value | `lst.count(x)` | |

### 📚 From class

````python
name = ["ali", "ahmed", "sana", 7, 10, 20, True]
name_2 = ["john", "mike", "ava"]

name.insert(3, "afraz")      # put "afraz" at index 3
print(name)

name.extend(name_2)           # add every item of name_2
print(name)

name.remove("afraz")          # remove by value
print(name)

name.pop()                    # remove the last item
print(name)

name.pop(0)                   # remove the item at index 0
print(name)

del name[3]                   # delete by position
print(name)

name.clear()                  # empty the list
print(name)
````

````python
numbers = [5, 2, 9, 1]

numbers.sort()                 # ascending by default
print(numbers)

numbers.sort(reverse=True)     # descending
print(numbers)

numbers.reverse()              # flips the current order
print(numbers)
````

````python
fruits = ["apple", "banana", "cherry", "banana"]
print(fruits.index("banana"))     # position of the FIRST match
print(fruits.count("banana"))     # how many times it appears
````

````python
print(fruits.index("mango"))      # "mango" isn't there
````

### 🆕 More examples

`append` adds **one** thing; `extend` adds **each item**. Compare:

````python
a = [1, 2]
a.append([3, 4])          # the whole list becomes ONE item
print(a)

b = [1, 2]
b.extend([3, 4])          # each item is added
print(b)

print([1, 2] + [3, 4])    # + joins two lists into a NEW list
print([0] * 5)            # * repeats
````

`sort()` changes the list and returns `None`. `sorted()` leaves the original alone and returns a new list. Mixing them up is a very common bug:

````python
data = [3, 1, 2]
result = data.sort()
print(result)             # None!  sort() changes data and returns nothing
print(data)

original = [3, 1, 2]
new_list = sorted(original)
print(original, new_list)
````

A tiny **shopping list** program using `append`, `remove` and `in`:

````python
shopping = ["milk", "eggs"]
shopping.append("bread")

item = "eggs"
if item in shopping:
    shopping.remove(item)
    print(f"Removed {item}")

print("Left to buy:", shopping)
````

## 3. Copying lists (a famous trap)

A variable is a **name** that points to a list. Writing `list2 = list1` does **not** copy the list. It just gives the *same list* a second name.

### 📚 From class

````python
list1 = [1, 2, 3]
list2 = list1            # NOT a copy: same list, two names
list2.append(4)
print(list1)             # list1 changed too!
print(list1 is list2)    # True: they are literally the same object
````

To make a real, independent copy, use one of these:

````python
thislist = ["apple", "banana", "cherry"]

mylist1 = thislist.copy()       # the .copy() method
mylist2 = list(thislist)        # the list() function
mylist3 = thislist[:]           # slice everything

thislist.append("mango")
print(thislist)
print(mylist1, mylist2, mylist3)
````

### 🆕 More examples

Why this matters: passing a list into a function lets the function change **your** list. (Compare with numbers, which can't be changed that way.)

````python
def add_bonus(scores):
    scores.append(100)        # changes the caller's list!

marks = [60, 70]
add_bonus(marks)
print(marks)
````

> 💡 A *shallow* copy copies the outer list only. If the list contains other lists (like a table), the inner lists are still shared. For that, use `import copy; copy.deepcopy(x)`.

## 4. Looping over lists, `enumerate`, `zip`

`for item in list:` visits each item in order.

````python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print("I like", fruit)
````

Need the **position** too? Use `enumerate`. Need to walk through **two lists together**? Use `zip`.

````python
for i, fruit in enumerate(fruits):
    print(i, fruit)

print()
for i, fruit in enumerate(fruits, start=1):      # start counting at 1
    print(f"{i}. {fruit}")
````

````python
names = ["Ali", "Sara", "Zain"]
marks = [85, 92, 70]

for n, m in zip(names, marks):
    print(f"{n} scored {m}")
````

**Pattern: finding the biggest item with a loop.** (Python has `max()`, but this shows how it works inside.)

````python
nums = [4, 17, 9, 23, 5]
biggest = nums[0]
for n in nums:
    if n > biggest:
        biggest = n
print("Biggest:", biggest)
````

**Pattern: filtering** (build a new list from the items that pass a test):

````python
nums = [4, 17, 9, 23, 5, 12]
evens = []
for n in nums:
    if n % 2 == 0:
        evens.append(n)
print(evens)
````

> ⚠️ **Don't add or remove items from a list while looping over it.** The loop gets confused. Build a new list instead.

## 5. List comprehensions

A **list comprehension** builds a new list in one line. It is the `for`-loop + `append` pattern squeezed together:

```
[ what_to_store    for item in something    if condition ]
```

### 📚 From class

````python
# the loop way
squares = []
for x in range(1, 6):
    squares.append(x ** 2)
print(squares)

# the comprehension way
squares = [x ** 2 for x in range(1, 6)]
print(squares)
````

Look at how the pieces move. The expression that was inside `.append(...)` goes to the **front**, and the `for` line stays exactly the same:

```
for x in range(1, 6):         [ x ** 2   for x in range(1, 6) ]
    squares.append(x ** 2)       ↑ what to store   ↑ the same for-line
```

### 🆕 More examples

Add an `if` at the end to **filter**. This is like SQL's `WHERE`:

````python
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune"]

short = [p for p in planets if len(p) < 6]
print(short)

loud = [p.upper() + "!" for p in planets if len(p) < 6]
print(loud)
````

````python
nums = [4, 17, 9, 23, 5, 12]

evens = [n for n in nums if n % 2 == 0]
doubled = [n * 2 for n in nums]
lengths = [len(word) for word in ["python", "is", "fun"]]
print(evens)
print(doubled)
print(lengths)
````

Count things with `sum` + a comprehension of `True`/`False` (since `True` counts as 1):

````python
temps = [31, 28, 35, 40, 22]
hot_days = sum([t > 30 for t in temps])
print("Days above 30:", hot_days)
````

> 💡 **Readability counts.** If a comprehension is hard to read, use a normal loop. Short and clear beats clever.

## 6. Strings: basics and slicing

A **string** is a sequence of characters. You can use single or double quotes. Strings work like lists for **indexing, slicing, `len`, `in` and looping**, but they are **immutable**.

### 📚 From class

````python
txt = "good things in life are free"
print("free" in txt)
print("aww" not in txt)

for letter in "banana":
    print(letter, end=" ")
````

````python
print("my name is 'afraz'")
print('my name is "afraz"')
print("my name is john" + "Hello")
print("my name is \n" * 3)
````

### Indexing, slicing, length

````python
planet = "Pluto"
print(planet[0])           # first character
print(planet[-3:])         # last three characters
print(planet[1:4])
print(len(planet))
print(planet[::-1])        # reversed
````

### Immutable: you can't change one character

````python
planet[0] = "B"
````

To "change" a string you build a **new** one:

````python
new_planet = "B" + planet[1:]
print(new_planet)
````

### 🆕 More examples

Escape characters:

| You type | You get |
|---|---|
| `\'` | a single quote |
| `\"` | a double quote |
| `\\` | a backslash |
| `\n` | a new line |
| `\t` | a tab |

````python
print("That's \"cool\"")
print("Name:\tAli\nAge:\t15")
````

A **palindrome** reads the same both ways. Slicing makes this one line:

````python
word = "madam"
print(word == word[::-1])
````

## 7. String methods

Strings have lots of useful methods. **None of them change the original** (strings are immutable); they all return a new string or a value.

| Method | What it does |
|---|---|
| `.upper()` / `.lower()` / `.title()` | change case |
| `.strip()` | remove spaces from both ends |
| `.replace(old, new)` | replace text |
| `.split(sep)` | string → **list** of pieces (default: split on whitespace) |
| `"sep".join(list)` | **list** → string |
| `.startswith(x)` / `.endswith(x)` | True / False |
| `.find(x)` / `.index(x)` | position of the first match (`find` gives `-1`; `index` raises an error) |
| `.count(x)` | how many times |
| `.isdigit()`, `.isalpha()`, `.isupper()` | what kind of characters |

### 📚 From class

````python
a = "Hello, Horld!"
print(a.replace("H", "J"))

a = "Hello, World!, John"
b = a.split(",")
print(b)
````

````python
claim = "Pluto is a planet!"
print(claim.upper())
print(claim.lower())
print(claim.index("plan"))
print(claim.startswith("Pluto"))
print(claim.endswith("planet"))       # False: it ends with "planet!"
````

### 🆕 More examples

`split()` and `join()` are opposites:

````python
sentence = "python is really fun"
words = sentence.split()                 # split on spaces
print(words)
print(len(words), "words")

print("-".join(words))
print(" ".join(w.capitalize() for w in words))

date = "1956-01-31"
year, month, day = date.split("-")       # unpacking the pieces
print(day + "/" + month + "/" + year)
````

````python
messy = "   hello there   "
print(f"[{messy}]")
print(f"[{messy.strip()}]")
````

````python
password = "Secret123"
print(password.isdigit(), password.isalpha(), password.isalnum())
print(any(ch.isdigit() for ch in password))     # does it contain a digit?
print(password.count("e"))
````

**Word counter**, a very common first text program:

````python
text = "the cat and the hat and the bat"
words = text.split()
print("Total words:", len(words))
print("The word 'the' appears", words.count("the"), "times")
````

**Cleaning input.** Users type messy answers. `strip().lower()` makes comparisons friendly:

````python
answer = input("Do you want to continue? (yes/no) ")
if answer.strip().lower() == "yes":
    print("Continuing...")
else:
    print("Stopping.")
````

## 8. Formatting strings (f-strings)

Joining text with `+` gets messy, and you must convert numbers with `str()`. **f-strings** are cleaner: put `f` before the quote and write expressions inside `{ }`.

You can control how a value looks after a colon:

| Format | Meaning | Example |
|---|---|---|
| `{x:.2f}` | 2 decimal places | `3.14159` → `3.14` |
| `{x:,}` | thousands separator | `52910390` → `52,910,390` |
| `{x:.1%}` | percentage | `0.256` → `25.6%` |
| `{x:>8}` / `{x:<8}` / `{x:^8}` | right / left / centre in 8 spaces | |
| `{x:05}` | pad with zeros to width 5 | `42` → `00042` |

### 📚 From class

````python
age = 36.69234
txt = f"My name is John, I am {age:.2f}"
print(txt)
````

### 🆕 More examples

````python
name = "Sara"
price = 1499.5
qty = 3

print(f"{name} bought {qty} items.")
print(f"Total: Rs {price * qty:,.2f}")
print(f"Discount: {0.15:.0%}")
print(f"Order #{42:05}")
````

A small **table** using alignment:

````python
items = [("Apple", 120), ("Watermelon", 450), ("Fig", 90)]
for name, cost in items:
    print(f"{name:<12}{cost:>6}")
````

Older style: `"{} is {}".format(a, b)`. You will see it in older code and in the Kaggle lessons, and it works the same way. Prefer f-strings in new code.

````python
print("{} weighs {:.1f} kg".format("The box", 2.456))
````

## 9. Tuples

A **tuple** is like a list that **cannot be changed** after you create it (it is *immutable*). It uses parentheses `( )`.

Why would you want that?
- **Safety:** data that shouldn't change (coordinates, RGB colours, a date).
- Functions use tuples to **return several values** at once.
- Tuples can be used as **dictionary keys**; lists cannot.

### 📚 From class

````python
coordinates = (33.6844, 73.0479)
print(coordinates)
print(type(coordinates))
````

> ⚠️ **One-item tuple gotcha:** the **comma** makes the tuple, not the parentheses.

````python
not_a_tuple = (5)
print(type(not_a_tuple))       # int!

actual_tuple = (5,)
print(type(actual_tuple))      # tuple: the comma matters
````

````python
nums = (1, 2, 3, 3, 4, 3, 5)
print(nums[0], nums[-1])
print(nums[:])
print(nums.count(3))           # how many 3s
print(nums.index(4))           # where is the first 4
````

````python
contact = ("afraz", "0322-4884403")
contact[1] = "0322-4884402"       # tuples can't be changed
````

If you need to "change" a tuple, convert to a list, change it, and convert back:

````python
x = list(contact)
x[1] = "0322-4884402"
contact = tuple(x)
print(contact)
````

### Unpacking
Put several names on the left to pull a tuple apart into variables. The number of names must match the number of items.

````python
isb = (33.6, 24.05)
lat, long = isb
print(lat)
print(long)

a, b, c = 1, 2, 3            # this is a tuple being unpacked!
print(a, b, c)

x, y = 5, 10
x, y = y, x                  # the famous swap
print(x, y)
````

### Tuples in a list: looping with unpacking
This is the **contact book** from class: a list where each item is a `(name, phone)` tuple. The `for name, phone in ...` line unpacks each tuple.

````python
contacts = [
    ("Ahmed", "0300-1234567"),
    ("Sara", "0311-9876543"),
    ("Zain", "0345-5551234"),
]

search_name = input("Enter a name to search: ")

for name, phone in contacts:
    if name == search_name:
        print(f"{name}'s number is {phone}")
        break
else:
    print("Contact not found.")      # runs only if the loop never hit "break"
````

### 🆕 More examples

**Returning multiple values** from a function. Python really returns one tuple and you unpack it:

````python
def min_and_max(numbers):
    return min(numbers), max(numbers)

result = min_and_max([4, 9, 2, 7])
print(result, type(result))

low, high = min_and_max([4, 9, 2, 7])
print(f"low={low}, high={high}")
````

````python
x = 0.125
numerator, denominator = x.as_integer_ratio()     # a built-in method returning a tuple
print(numerator, denominator)
````

`enumerate` and `zip` (from section 4) give you tuples too. That's why `for i, fruit in enumerate(...)` works.

**Tuple as a dictionary key** (you'll meet dictionaries in the next notebook):

````python
city_names = {(33.68, 73.05): "Islamabad", (31.52, 74.35): "Lahore"}
print(city_names[(31.52, 74.35)])
````

## 10. Choosing between list, string and tuple

| Question | Use |
|---|---|
| A collection that will grow, shrink or be re-ordered? | **list** |
| Fixed group of related values (a point, a record, an RGB colour)? | **tuple** |
| Text? | **string** |
| Do I need to look things up by a *name* rather than a position? | a **dictionary** (next notebook) |
| Do I need to remove duplicates or test membership very fast? | a **set** (next notebook) |

All three support: indexing `x[i]`, slicing `x[a:b]`, `len(x)`, `in`, and looping with `for`.
