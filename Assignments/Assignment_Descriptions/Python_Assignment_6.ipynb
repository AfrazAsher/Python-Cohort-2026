### Question 4 — Predict the output

```python
class A:
    def hello(self):
        return "A"

    def greet(self):
        return "Hello from " + self.hello()

class B(A):
    def hello(self):
        return "B"

class C(B):
    def hello(self):
        return "C+" + super().hello()

for obj in [A(), B(), C()]:
    print(obj.greet())
```
