## 🔁 Recursion
*A function that solves a problem by calling itself on a smaller version of the same problem. Every function below needs a clear base case — without one, it will crash with a `RecursionError`.*

### Question 7 — Factorial

Write a recursive function `factorial(n)` that returns `n!` (n factorial), where `factorial(5) = 5 * 4 * 3 * 2 * 1 = 120`.

Think carefully about the base case: what should `factorial(0)` or `factorial(1)` return, without making any further recursive call?


### Question 8 — Sum of digits

Write a recursive function `sum_of_digits(n)` that adds up all the digits of a positive integer. For example, `sum_of_digits(1234)` should return `10` (since `1 + 2 + 3 + 4 = 10`).

Hint: `n % 10` gives you the last digit, and `n // 10` gives you the number with that last digit removed — this is how you shrink the problem on each call. What should the base case be, once there's only one digit left?


### Question 9 — Power function

Write a recursive function `power(base, exponent)` that calculates `base` raised to `exponent`, **without using the `**` operator or the built-in `pow()`**.

Think about the base case first: what is any number raised to the power of `0`? Then think about the recursive case: `power(base, exponent)` can be written in terms of `power(base, exponent - 1)`.


### Question 10 — Reverse a string

Write a recursive function `reverse_string(text)` that returns the string reversed, without using slicing (`text[::-1]`) or any built-in reverse method.

Think about the base case: what is the reverse of an empty string, or a string with just one character? For the recursive case, think about what the first character of the string should do relative to the reverse of everything *after* it.


### Question 11 — Recursive sum of a list
