### Question 21 — Ternary expression — username or guest

You're given a variable `username`, which might be an empty string `""`. Using a **ternary expression**, print the username if it exists, otherwise print `"Guest"`.

A ternary expression is a **shorter way to write an if/else** that produces a value in a single line. The pattern is:

```python
value_if_true if condition else value_if_false
```

For example, this normal if/else:

```python
if age >= 18:
    status = "Adult"
else:
    status = "Minor"
```

can be written as a ternary in one line:

```python
status = "Adult" if age >= 18 else "Minor"
```

Both versions do exactly the same thing — the ternary is just shorter. For this question, an empty string `""` is treated as `False` in a condition, so `if username` is `True` only when `username` actually has text in it.
